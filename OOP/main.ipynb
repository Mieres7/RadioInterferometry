```python
def generar_datos_simulados(n_antenas=5):
    return [
        {'id': i, 'longitude': -67.0 + i*0.01, 'latitude': -23.0 + i*0.01} 
        for i in range(n_antenas)
    ]

def experimento():
    datos = generar_datos_simulados(100)
    n_iteraciones = 100 # Reducido para el ejemplo visual, aumentar para carga real
    
    print("\n--- INICIO EXPERIMENTO: Con cerrar() explícito ---")
    start_explicit = time.perf_counter()
    for _ in range(n_iteraciones):
        interf = Interferometer(datos)
        interf.reserve_calibration()
        interf.close() # Limpieza explícita
    end_explicit = time.perf_counter()

    print("\n--- INICIO EXPERIMENTO: Sin cerrar() (Dependiendo de GC) ---")
    start_implicit = time.perf_counter()
    for _ in range(n_iteraciones):
        interf = Interferometer(datos)
        interf.reserve_calibration()
        # No llamamos a cerrar(), dejamos que el scope limpie la variable 'interf'
    end_implicit = time.perf_counter()

    print("\n" + "="*40)
    print(f"Tiempo con cerrar() explícito: {end_explicit - start_explicit:.4f} seg")
    print(f"Tiempo implícito (GC):       {end_implicit - start_implicit:.4f} seg")
    print("="*40)

# Para correr el experimento, descomenta la siguiente línea:
experimento()
```
